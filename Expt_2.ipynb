```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error,r2_score
```

```python
url= "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/mpg.csv"

df=pd.read_csv(url)

df.head()
```

```python
x=df[["displacement"]]
y=df["mpg"]

print("Missing values in ", x.isnull().sum())
print("Missing values in mpg:",y.isnull().sum())
```

```python
#to remove rows missing value 
data=df[['displacement','mpg']].dropna()
x=data[['displacement']]
y=data['mpg']

print("Feature:")
print(x.head())
print("\n Target:")
print(y.head)
```

```python
plt.figure(figsize=(8,5))
plt.scatter(x,y)
plt.xlabel("Engine Dsiplacement")
plt.ylabel("Miles Per Gallon")
plt.title("Engine Displacement vs MPG")
plt.grid(True)
plt.show()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

print("Training samples:", len(x_train))
print("Testing Samples:", len(x_test))
```
